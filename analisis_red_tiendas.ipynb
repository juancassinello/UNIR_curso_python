```python
# 2. Importar las librerías necesarias:
import numpy as np
import pandas as pd
```

```python
# 3. Cargar los datos (lectura y procesamiento de datos - Pandas):
# Limpia los datos eliminando filas con valores nulos utilizando el método dropna(). Esto te permitirá trabajar solo con datos válidos.

# SALES
df_sales = pd.read_csv('1756803113331-sales.csv')
df_sales.dropna()
print(df_sales.head())

# INVENTORIES
df_inventories = pd.read_csv('1756803222042-inventories.csv')
df_inventories.dropna()
print(df_inventories.head())

# SATISFACTION
df_satisfaction = pd.read_csv('1756803304185-satisfaction.csv')
df_satisfaction.dropna()
print(df_satisfaction.head())

# MERGE SALES + INVENTORIES
df_sales_inventories = pd.merge(df_sales, df_inventories, on=['ID_Tienda', 'Producto'])
df_sales_inventories.head()
```

```python
# ----------------------------------
#  4. Exploración de datos (Pandas)
# ----------------------------------

# 4.1.1 Calcula las CANTIDADES VENDIDAS TOTALES por PRODUCTO.
cantidades_totales_producto = df_sales_inventories.groupby('Producto')['Cantidad_Vendida'].sum()
#print(f"Las CANTIDADES VENDIDAS TOTALES por PRODUCTO son {cantidades_totales_producto}")

# 4.1.2 Calcula las CANTIDADES VENDIDAS TOTALES por TIENDA.
cantidades_totales_tienda = df_sales_inventories.groupby('ID_Tienda')['Cantidad_Vendida'].sum()
#print(f"Las CANTIDADES VENDIDAS TOTALES por TIENDA son {cantidades_totales_tienda}")

# 4.2 Calcula los INGRESOS TOTALES por TIENDA: multiplicando la CANTIDAD VENDIDA por el PRECIO.
df_sales_inventories['Ingresos_Totales_Tienda'] = df_sales_inventories['Cantidad_Vendida'] * df_sales_inventories['Precio_Unitario']
#df_sales_inventories.head()
print(f"Los INGRESOS TOTALES por TIENDA son: {df_sales_inventories.groupby('ID_Tienda')['Ingresos_Totales_Tienda'].sum()}")

# 4.3 Genera un resumen estadístico de las VENTAS utilizando el método describe() para obtener la media, mediana y otras métricas clave.
resumen_estadistico = df_sales_inventories['Cantidad_Vendida'].describe()
print(resumen_estadistico)

# 4.4 Si los PRODUCTOS están clasificados por CATEGORÍAS: calcula el promedio de VENTAS por TIENDA y CATEGORÍA de productos
# NO VEO QUE ESTÉN CATEGORIZADOS POR CATEGORÍAS

```

```python
# ----------------------------------
# 5. Análisis de inventarios (Pandas)
# ----------------------------------

# 5.1 Calcula la ROTACIÓN de INVENTARIOS para cada TIENDA: dividiendo las VENTAS TOTALES por el STOCK disponible de cada producto.
df_sales_inventories['Rotacion'] = df_sales_inventories['Cantidad_Vendida'] / df_sales_inventories['Stock_Disponible'] * 100
df_sales_inventories.head()

# 5.2 Almacena los resultados en una nueva columna dentro del DataFrame de inventarios.
# HECHO

#5.3 Filtra y muestra las tiendas con niveles críticos de inventario:
# aquellas tiendas donde el porcentaje de productos vendidos sea menor al 10% del stock disponible.
tiendas_criticas = df_sales_inventories[df_sales_inventories['Rotacion'] < 10]
print(tiendas_criticas)
```
